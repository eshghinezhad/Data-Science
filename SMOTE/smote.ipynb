Python Demo With SMOTE (Best Practice)
1\Install/import (if needed)

```python
# If you don't have it installed:
# pip install imbalanced-learn

from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix

from imblearn.over_sampling import SMOTE
from imblearn.pipeline import Pipeline
```

2) Create imbalanced data

```python
X, y = make_classification(
    n_samples=1000,
    n_features=2,
    n_informative=2,
    n_redundant=0,
    weights=[0.99, 0.01],
    random_state=42
)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
```

3) Baseline model (no fix)

```python
baseline = LogisticRegression()
baseline.fit(X_train, y_train)

pred_base = baseline.predict(X_test)
print("BASELINE (no fix)")
print(confusion_matrix(y_test, pred_base))
print(classification_report(y_test, pred_base))
```

4) Add SMOTE using a Pipeline (recommended)

```python
smote_clf = Pipeline(steps=[
    ("smote", SMOTE(random_state=42, k_neighbors=3)),
    ("model", LogisticRegression())
])

smote_clf.fit(X_train, y_train)
pred_smote = smote_clf.predict(X_test)

print("SMOTE + Logistic Regression")
print(confusion_matrix(y_test, pred_smote))
print(classification_report(y_test, pred_smote))
```
